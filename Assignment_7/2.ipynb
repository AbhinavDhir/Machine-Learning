```python
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score 

data = load_breast_cancer() 
x = data.data
y = data.target

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(f"Dataset features shape: {x.shape[1]}") 
print(f"Training set size: {x_train.shape[0]}, Test set size: {x_test.shape[0]}")


print("Model 1 : without feature scaling") 

model_no_scaling = SVC(kernel='rbf', random_state=42)
model_no_scaling.fit(x_train, y_train)

y_train_pred_1 = model_no_scaling.predict(x_train)
train_accuracy_1 = accuracy_score(y_train, y_train_pred_1)

y_test_pred_1 = model_no_scaling.predict(x_test)
test_accuracy_1 = accuracy_score(y_test, y_test_pred_1) 

print(f"Training Accuracy (no scaling): {train_accuracy_1:.4f}")
print(f"Test Accuracy (no scaling): {test_accuracy_1:.4f}")

print("Model 2 : with feature scaling") 

scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

model_with_scaling = SVC(kernel='rbf', random_state=42)
model_with_scaling.fit(x_train_scaled, y_train)

y_train_pred_2 = model_with_scaling.predict(x_train_scaled)
train_accuracy_2 = accuracy_score(y_train, y_train_pred_2)

y_test_pred_2 = model_with_scaling.predict(x_test_scaled)
test_accuracy_2 = accuracy_score(y_test, y_test_pred_2)

print(f"Training Accuracy (with scaling): {train_accuracy_2:.4f}")
print(f"Test Accuracy (with scaling): {test_accuracy_2:.4f}")

results = { 
    "Model": ["Without Scaling", "With Scaling"],
    "Training Accuracy": [train_accuracy_1, train_accuracy_2],
    "Test Accuracy": [test_accuracy_1, test_accuracy_2]
}

df_results = pd.DataFrame(results)
print("\nComparison of Model Performance:")
print(df_results.to_string(index=False))

```
